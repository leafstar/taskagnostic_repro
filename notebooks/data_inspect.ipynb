# Camargo Trial Sanity Check

Set `trial_path` to one local `.mat` file, inspect available fields, then load the configured input/target arrays.

```python
import os
from pathlib import Path
import sys

ROOT = Path('..').resolve()
sys.path.insert(0, str(ROOT))

from src.utils import load_config
from data.camargo_loader import inspect_mat, load_trial

cfg = load_config(ROOT / 'configs/default.yaml')
data_root = Path(os.environ.get('CAMARGO_DATA_ROOT', '../data')).expanduser()
trial_path = data_root / 'raw' / 'AB01' / 'path' / 'to' / 'trial.mat'
```

```python
inspect_mat(trial_path, max_items=120)
```

```python
trial = load_trial(
    trial_path,
    feature_patterns=cfg['data']['feature_patterns'],
    target_patterns=cfg['data']['target_patterns'],
    target_hz=cfg['data']['sample_rate_hz'],
    data_root=cfg['data']['root'],
)
trial.x.shape, trial.y.shape, trial.feature_names[:10], trial.target_names
```

```python
import matplotlib.pyplot as plt

fig, axs = plt.subplots(2, 1, figsize=(12, 6), sharex=True)
axs[0].plot(trial.x[:, :min(8, trial.x.shape[1])])
axs[0].set_title('First input channels')
axs[1].plot(trial.y)
axs[1].legend(['hip moment', 'knee moment'])
axs[1].set_title('Targets, Nm/kg if mass was found')
plt.tight_layout()
```
